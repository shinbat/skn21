
# Pandas 정렬과 집계 완전 정리

## 🎯 Pandas의 핵심 역할

pandas는 **"데이터 분석을 위한 Excel의 Python 버전"**입니다.
- **데이터 정리**: 정렬, 필터링, 변환
- **데이터 집계**: 그룹별 통계, 요약
- **데이터 분석**: 패턴 발견, 인사이트 도출

---

## 🔄 1. 정렬 (Sorting)

### 기억해야 할 포인트

```python
# 인덱스/컬럼 이름으로 정렬
df.sort_index()          # 행 이름으로 정렬
df.sort_index(axis=1)    # 열 이름으로 정렬

# 값으로 정렬
df.sort_values('컬럼명')           # 단일 컬럼
df.sort_values(['컬럼1', '컬럼2']) # 다중 컬럼
```

### 💡 이해 팁
- `sort_index()` = **이름표 정렬** (행/열 이름)
- `sort_values()` = **내용물 정렬** (실제 데이터 값)
- `ascending=False`로 내림차순
- **결측치는 항상 마지막에 위치**

---

## 📊 2. 집계 (Aggregation)

### 기본 통계 메서드 (외워두면 좋은 것들)

```python
# 숫자형 데이터
df.sum()      # 합계
df.mean()     # 평균 ⭐
df.median()   # 중앙값
df.std()      # 표준편차
df.min()      # 최솟값
df.max()      # 최댓값
df.count()    # 개수 (결측치 제외) ⭐

# 문자형도 가능
df.nunique()  # 고유값 개수 ⭐
df.mode()     # 최빈값
```

### 💡 이해 팁
- **DataFrame에 적용 → 컬럼별로 계산**
- `axis=0` (기본): 세로방향 (컬럼별)
- `axis=1`: 가로방향 (행별)
- `skipna=True` (기본): 결측치 무시

---

## 🎯 3. agg() - 다양한 집계 한번에

### 기억해야 할 패턴

```python
# 여러 집계함수 동시 적용
df['컬럼'].agg(['mean', 'std', 'min', 'max'])

# 컬럼별로 다른 집계
df.agg({
    'age': ['mean', 'std'],
    'salary': ['sum', 'count']
})

# 사용자 정의 함수
def 범위(x):
    return x.max() - x.min()

df['컬럼'].agg(범위)
```

---

## 🔢 4. Groupby - 그룹별 집계 (⭐ 가장 중요)

### 핵심 개념

```python
# 기본 패턴: "~별 ~를 구하기"
df.groupby('그룹컬럼')['집계컬럼'].집계함수()

# 예시
df.groupby('성별')['나이'].mean()        # 성별 평균나이
df.groupby('부서')['급여'].sum()         # 부서별 급여합계
df.groupby(['지역', '성별'])['점수'].mean()  # 지역+성별별 평균점수
```

### 💡 이해 팁
- **Excel의 피벗테이블과 같은 개념**
- 범주형 컬럼으로 그룹을 만들고
- 수치형 컬럼을 집계

### groupby 결과 활용

```python
result = df.groupby('항공사')['취소건수'].sum()
result[result > 100]  # 취소건수 100건 이상만
```

---

## 📋 5. Pivot Table - 보기 좋은 집계표

### 언제 사용하나?
- **여러 컬럼으로 그룹핑할 때**
- **결과를 표 형태로 예쁘게 보고 싶을 때**

```python
df.pivot_table(
    index='행에올컬럼',     # 행 기준
    columns='열에올컬럼',   # 열 기준  
    values='집계할컬럼',    # 집계 대상
    aggfunc='mean'        # 집계함수
)
```

### 💡 이해 팁
- **groupby의 예쁜 버전**
- `margins=True`로 총계 추가 가능

---

## ✂️ 6. Cut - 수치형을 범주형으로 변환

### 언제 사용하나?
- 나이 → 연령대
- 점수 → 등급
- 키 → 크기 그룹

```python
# 자동으로 구간 나누기
pd.cut(df['나이'], bins=3, labels=['청년', '중년', '노년'])

# 원하는 구간으로 나누기  
pd.cut(df['점수'], bins=[0, 60, 80, 100], labels=['하', '중', '상'])
```

### 💡 이해 팁
- **연속형 데이터를 그룹으로 만들 때**
- `right=True`(기본): 오른쪽 끝점 포함
- `bins`에 숫자 → 등간격, 리스트 → 원하는 구간

---

## 🚀 7. Apply - 일괄 처리의 핵심

### 언제 사용하나?
- **모든 행/열/원소에 같은 작업**
- 반복문 대신 사용

```python
# Series에 적용 (각 원소마다)
df['컬럼'].apply(함수)

# DataFrame에 적용
df.apply(함수, axis=0)  # 각 컬럼마다
df.apply(함수, axis=1)  # 각 행마다
```

### 💡 이해 팁
- **반복문을 쓰고 싶을 때 apply를 먼저 생각**
- lambda 함수와 자주 함께 사용
- 결측치 처리도 잊지 말기

### 1. 자주 쓰는 패턴 외우기

```python
# 그룹별 집계의 황금패턴
df.groupby('그룹')['대상'].집계함수()

# 조건 필터링
df[df['컬럼'] > 값]

# 정렬 후 상위/하위
df.sort_values('컬럼').head(10)
```

### 3. 결측치 처리 기본
- `skipna=True` (기본): 결측치 무시
- `dropna()`: 결측치 제거
- `fillna()`: 결측치 채우기

## 🎪 pandas의 진짜 파워

### 데이터 분석의 3단계
1. **정리**: 정렬, 필터링, 결측치 처리
2. **요약**: groupby, 집계함수로 패턴 파악
3. **변환**: apply, cut으로 새로운 컬럼 생성

### 실무에서 이렇게 사용

```python
# 1. 데이터 로드 및 확인
df = pd.read_csv('data.csv')
df.info()
df.describe()

# 2. 그룹별 분석
df.groupby('카테고리')['매출'].agg(['sum', 'mean', 'count'])

# 3. 상위/하위 분석
top10 = df.groupby('지역')['매출'].sum().sort_values(ascending=False).head(10)

# 4. 새로운 파생변수 생성
df['매출등급'] = pd.cut(df['매출'], bins=3, labels=['하', '중', '상'])

# 5. 최종 요약표
결과 = df.pivot_table(
    index='지역', 
    columns='매출등급', 
    values='고객수', 
    aggfunc='sum'
)
```
